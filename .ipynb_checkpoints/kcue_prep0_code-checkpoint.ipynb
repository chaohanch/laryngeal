## Visualizing spectrum of data

```python
## parameters
# frequency of the eeg recording for audio downsampling
# eeg_freq = 500


# directory
input_dir = os.getcwd() + '/../raw eeg/'
output_dir = os.getcwd() + '/../raw eeg/'

# file name
file = 'CUE2024_0008_VOT_corr.fif'

# read in file
raw = mne.io.mne.io.read_raw_curry(input_dir + file, preload = True)

# drop trigger channel
# raw.drop_channels(['Trigger'])

# check information
print(raw.info)

# plot spectrum of raw EEG
# raw.compute_psd().plot()

# if needed, plot a spectrum for each channel
for ch_ind in raw.info['ch_names']:
    print('=====================')
    print(ch_ind)
    raw.compute_psd(picks = ch_ind).plot(show = False, amplitude=False)
    plt.show()
```

## Save for EEGLAB

```python
# directory
input_dir = os.getcwd() + '/../preprocessed/3_ica/'
output_dir = os.getcwd() + '/../preprocessed/test/'

# file name
file = 'CUE2024_0008_VOT_corr_prep_ica.fif'

# read in file
raw = mne.io.read_raw_fif(input_dir + file, preload = True)

# save
mne.export.export_raw(output_dir + "CUE2024_0008_VOT_corr_prep_ica.set", 
                      raw, fmt='auto', physical_range='auto', add_ch_type=False, overwrite=True, verbose=None)
```
