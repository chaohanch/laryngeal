```python
# libraries
import os
import mne
import numpy as np

# import pyprep
from pyprep.prep_pipeline import PrepPipeline

"""
Overview:
1. preprocessing
    1.1. filtering
    1.2. resampling
    1.4. remove line noise
    1.5. Bad channel detection & repairing
    1.6. add back reference channel TP9
"""
```

```python
#### parameters ####

# set directory
input_dir = os.getcwd() + '/../preprocessing/1_trigger_latency_corrected/'
output_dir = os.getcwd() + '/../preprocessing/1_trigger_latency_corrected/'

# # filter cutoff frequencies (low/high)
# f_low = 0.1
# f_high = 40

# # resampling frequency
# f_res = 250

# # line frequency
# line_freq = 60


# interpolation method
# method=dict(eeg="spline")
```

```python
#####################################################
#### Preprocessing (filtering, resampling, bad channel detection/interpoloation, re-reference) ####
#####################################################


file = 'CUE2024_0008_VOT_corr.fif'
        
# read in file
raw = mne.io.read_raw_fif(input_dir + file, preload=True)

# get channel index
chan_idx = np.isin(raw.info['ch_names'], 'C4')

# extract data
data = raw.get_data()

# change the C4 data to 0
data[chan_idx,:] = np.zeros(data[chan_idx,:].shape)

# replace data
raw._data = data

# save
raw.save(output_dir + file.split('.')[0]+ '.fif', overwrite = True)

# release memory
del raw
```
